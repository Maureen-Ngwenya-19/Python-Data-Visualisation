# Data Visualisation

 Important libraries we use 


 - Pandas : for wokring with tables and dataframes

 - Numpy : for generating random numbers and mathematical operations

 - matplotlib : for creating visualisation

```python
## STEP 0 : Import libraries

import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
### Step 1 : creating out first line plot 

df = pd.DataFrame(np.random.randint(0,100,size =(10)))

df
```

```python
# creating line plot 

plt.plot(df)  # if you dont define your graph it automatically draws a line plot 

plt.xlabel("Time")
plt.ylabel("Values")
plt.title("Basic Line Plot")
plt.show()
```

```python
## Step 2 : More complex line plogts with multiple columns 



# create a dataframe with 10 rowa and 4 columns (A,B,C,D)

df = pd.DataFrame(
    np.random.randint(0, 100, size=(10, 4)),
    columns=list("ABCD")
)

df
```

```python
df.plot()

plt.title("Multiple Line plot")
plt.xlabel("Time")
plt.ylabel("Values")
plt.show()
```

```python
# Step 3 : working with multiple plots () [subplots()]

# sometimes we want to make comparisons and we can show these charts together in a subplot

df1 = pd.DataFrame(np.random.randint(0,10 , size = (10)))
df2 = pd.DataFrame(np.random.randint(0,10 , size = (10)))


# create a figure with 2 subplots  ( 2 rows , 1 column)

fix, axes  = plt.subplots(nrows = 2 , ncols =1 , figsize = (8,6))
```

```python
fig,axes = plt.subplots(nrows =2 , ncols= 1 , figsize =(8,6))

# plot df1 on the first subplot(axes(0))

df1.plot(ax = axes[0] , legend = True)
axes[0].set_title('Dataframe 1')

# Plt df2 on the second subplot(axes[1])

df2.plot(ax=axes[1] , legend = True)
axes[1].set_title('Dataframe 2')

plt.show()
```

```python
# removing the legend

fig,axes = plt.subplots(nrows =2 , ncols= 1 , figsize =(8,6))

# plot df1 on the first subplot(axes(0))

df1.plot(ax = axes[0] , legend = False)
axes[0].set_title('Dataframe 1')

# Plt df2 on the second subplot(axes[1])

df2.plot(ax=axes[1] , legend = False)
axes[1].set_title('Dataframe 2')

plt.show()
```

```python
## STEP 4 : BAR PLOTS


## Vertical bar plots 


df1 = pd.DataFrame(np.random.rand(12) , index = list('abcdefghijkl'))
df1
```

```python
df1.plot.bar()

plt.title("Vertical bar plot")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

```python
### Horizontal bar plots

df1.plot.barh()  # horizontal bar plots

plt.title("Vertical bar plot")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

```python
# Grouped bar plots

df1 = pd.DataFrame(np.random.rand(4,4),
                   index = ['1','2','3','4'],
                   columns = pd.Index(['A','B','C','D'] , name ='Categories'))

df1
```

```python
# Plot a group bar plot

df1.plot.bar()

plt.show()
```

```python
# Plot a group bar plot

df1.plot.barh()

plt.show()
```

```python
### stacked bar chart

df1.plot.barh(stacked =True)

plt.show()
```

```python
### stacked bar chart

df1.plot.bar(stacked =True)

plt.show()
```

```python
### step 5: scatter plot (show re lationship between two variables)

df2 = pd.DataFrame(
    np.random.randint(0, 20, size=(30, 2)),
    columns=list('xy'))

import seaborn as sns    # for statistical measures


# create a scatter plot 

sns.regplot(x='x', y='y' , data =df2)
plt.show()
```

```python
## step 6 : Pair plot (quickets way to get insights)


df3 = pd.DataFrame(np.random.randint(0,20 , size=(30,4)) , columns=list('WXYZ'))

sns.pairplot(df3)
plt.show()

# shows the insight for all relationship (x,y) ,(y,x) and (z,z){ this one you dont need since it shows z distribution , shows a histogram }
```

```python
# pair plot with density diagonal

sns.pairplot(df3 , diag_kind='kde')  # kde = kernel density estimate ( gives area under the curve)
```

```python
#  step 7 :histogram ( gives distribution of your variables in a dataset)

# create  an array with repeated values 

myarray = np.array([7,7,0,2,6,3,6,1,5,2,1,5,3,5]) # shows frequency in terms of repeation


plt.hist(myarray , bins=[0,2,4,6,8] ,edgecolor ='black')

plt.show()
```
